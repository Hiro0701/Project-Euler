```python
def sum_of_squares(n: int) -> int:
    return sum(
        [
            i ** 2 for i in range(1, n+1)
        ]
    )
```

```python
def square_of_sum(n: int) -> int:
    return sum(
        [
            i for i in range(1, n+1)
        ]
    ) ** 2
```

```python
print(square_of_sum(10) - sum_of_squares(10))
```

```python
print(square_of_sum(100) - sum_of_squares(100))
```

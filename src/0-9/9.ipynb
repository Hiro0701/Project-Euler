```python
from itertools import combinations

# 제너레이터화 하면 좋을듯?
def find_special_triplet(n: int) -> int:
    square_list = [
        i ** 2 for i in range(1, n-1)
    ]

    for i in range(n-1, 1, -1):
        combination_list = list(combinations(list(range(n-1-i, 1, -1)), 2))
        for j, k in combination_list:
            if i ** 2 == (j ** 2 + k ** 2):
                if (i + j + k) == n:
                    return (i, j, k)
```

```python
triplet = find_special_triplet(1000)
```

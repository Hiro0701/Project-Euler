```python
def find_largest_palindrome(n: int) -> (int, int, int):
    assert n > 2

    largest_palindrome = 0
    current_product = 0

    for a in range(10 ** (n - 1), 10 ** n):
        for b in range(10 ** (n - 1), 10 ** n):
            current_product = a * b
            if str(current_product) == str(current_product)[::-1]:
                if current_product > largest_palindrome:
                    largest_palindrome = current_product

    return a, b, largest_palindrome
```

```python
a, b, largest_palindrome = find_largest_palindrome(3)

print(a, b, largest_palindrome)
```

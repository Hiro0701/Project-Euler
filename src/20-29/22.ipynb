```python
def calc_alphabetical_value(name: str) -> int:
    return sum(
        [
            (ord(i.lower()) - 96) for i in name
        ]
    )
```

```python
with open('0022_names.txt', 'r') as f:
    names = f.read().split(",")
    names = [
        name.replace('"', '') for name in names
    ]
    names = sorted(names)

    score_sum = 0
    for i, name in enumerate(names):
        score_sum += calc_alphabetical_value(name) * (i + 1)

    print(score_sum)
```

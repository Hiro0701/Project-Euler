```python
def find_proper_divisors(n: int) -> list:
    if n == 1:
        return [1]

    divisor_list = []
    for i in range(1, n):
        if n % i == 0:
            divisor_list.append(i)

    return divisor_list
```

```python
def find_amicable_numbers(n: int) -> dict:
    divisor_sum_dict = {}
    amicable_number_dict = {}

    for i in range(2, n+1):
        divisor_sum_dict[i] = sum(find_proper_divisors(i))

    for i in divisor_sum_dict:
        if (divisor_sum_dict[i] <= n + 1) and (divisor_sum_dict[i] > 1) and (i != divisor_sum_dict[i]):
            if divisor_sum_dict[divisor_sum_dict[i]] == i:
                if not divisor_sum_dict[i] in amicable_number_dict:
                    amicable_number_dict[i] = divisor_sum_dict[i]

    return amicable_number_dict
```

```python
amicable_numbers_10000 = find_amicable_numbers(10000)

amicable_numbers_10000
```

```python
def find_longest_reciprocal_cycle(n: int) -> tuple:
    longest_reciprocal_tuple = (0, 0)

    for i in range(2, n+1):
        remainder_list = [1]
        current_idx = 0

        while True:
            current_idx += 1
            current_remainder = remainder_list[current_idx - 1] * 10 % i

            #순환마디를 찾을 때
            if current_remainder in remainder_list:
                if len(remainder_list) > longest_reciprocal_tuple[1]:
                    longest_reciprocal_tuple = (i, len(remainder_list))
                break

            #나머지가 0이 될 때
            elif current_remainder == 0:
                break

            #그 외
            else:
                remainder_list.append(current_remainder)

    return longest_reciprocal_tuple
```

```python
longest_reciprocal_cycle_1000 = find_longest_reciprocal_cycle(1000)

longest_reciprocal_cycle_1000
```

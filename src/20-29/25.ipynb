```python
def find_fibonacci_number_digit(n: int) -> tuple:
    fibonacci_tuple = (1, 1)
    idx = 3

    while True:
        fibonacci_tuple = (fibonacci_tuple[1], fibonacci_tuple[0] + fibonacci_tuple[1])
        if len(str(fibonacci_tuple[1])) > 999:
            return idx
        idx += 1
```

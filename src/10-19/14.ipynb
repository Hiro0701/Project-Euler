```python
from tqdm import tqdm
```

```python
def find_longest_collatz_seq(n:int) -> tuple:
    assert n >= 2

    collatz_len_dict = {}
    collatz_dict = {}
    
    with tqdm() as pbar:
        for i in range(2, n+1):
            current_len = 0
            next_num = 0
            current_num = i

            while next_num != 1:
                if current_num in collatz_dict:
                    next_num = collatz_dict[current_num]
                else:
                    if current_num % 2 == 0:
                        next_num = current_num // 2
                    else:
                        next_num = 3 * current_num + 1
                    collatz_dict[current_num] = next_num
                current_len += 1
                current_num = next_num
            
            collatz_len_dict[i] = current_len

        longest_collatz_n = max(collatz_len_dict, key=collatz_len_dict.get)

        return longest_collatz_n, collatz_len_dict[longest_collatz_n]
```

```python
longest_collatz_1000000 = find_longest_collatz_seq(1000000)

longest_collatz_1000000
```

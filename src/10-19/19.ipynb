```python
from itertools import count
```

```python
year_dict = {
    1: 31,
    2: 28,
    3: 31,
    4: 30,
    5: 31,
    6: 30,
    7: 31,
    8: 31,
    9: 30,
    10: 31,
    11: 30,
    12: 31,
}
```

```python
# TODO 1900년 이전도 검색 가능하게 하는 것
def counting_days(target_day: int, start_ymd: list, end_ymd: list, condition_ymd: list) -> int:
    current_day = 0
    current_date = 1
    current_month = 1
    current_year = 1900

    start_date = start_ymd[2]
    start_month = start_ymd[1]
    start_year = start_ymd[0]

    end_date = end_ymd[2]
    end_month = end_ymd[1]
    end_year = end_ymd[0]

    condition_date = condition_ymd[2]
    condition_month = condition_ymd[1]
    condition_year = condition_ymd[0]

    is_in_period = False
    day_count = 0

    for i in count(1):
        if current_date == start_date and \
            current_month == start_month and \
            current_year == start_year:
            is_in_period = True

        if current_day == target_day:
            if is_in_period == True:
                if (not condition_date) or (current_date == condition_date):
                    if (not condition_month) or (current_month == condition_month):
                        if (not condition_year) or (current_year == condition_year):
                            day_count += 1

        if current_date == end_date and \
            current_month == end_month and \
            current_year == end_year:
            return day_count
        
        if current_date == year_dict[current_month]:
            if current_month == 2:
                if (current_year % 4 == 0) and (current_year % 100 != 0):
                    current_date += 1
                else:
                    if current_year % 400 == 0:
                        current_date += 1
                    else:
                        current_month += 1
                        current_date = 1
            else:
                if current_month != 12:
                    current_month += 1
                    current_date = 1
                else:
                    current_month = 1
                    current_year += 1
                    current_date = 1
        else:
            if current_date == 29 and current_month == 2:
                current_month += 1
                current_date = 1
            else:
                current_date += 1

        current_day = i % 7
```

```python
sunday_counts = counting_days(
    6,
    [
        1901,
        1,
        1,
    ],
    [
        2000,
        12,
        31,
    ],
    [
        None,
        None,
        1,
    ],
)

sunday_counts
```

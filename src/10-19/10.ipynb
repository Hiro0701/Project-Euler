```python
# 에라토스테네스의 체를 이용한 소수 판별 함수

import math

def find_prime_numbers_by_n(n: int) -> list:
    sqrt_n = int(math.sqrt(n))

    prime_number_list = list(range(2, n+1))

    for i in range(2, sqrt_n+1):
        if i in prime_number_list:
            prime_number_list = [
                x for x in prime_number_list if (x % i != 0 or x == i)
            ]

    return prime_number_list
```

```python
prime_numbers_by_2000000 = find_prime_numbers_by_n(2000000)

sum(prime_numbers_by_2000000)
```

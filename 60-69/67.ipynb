```python
# From Problem 18
def find_max_path_sum(path_list: list) -> tuple:
    path_len = len(path_list)
    best_paths_dict = {
        0: path_list[0],
    }

    for i in range(path_len - 1):
        current_best_paths = best_paths_dict[i]
        next_path_candidates = path_list[i+1]

        next_best_paths = []
        best_sub_path = 0
        for j, path in enumerate(current_best_paths):
            if j == 0 and len(current_best_paths) != 1:
                next_best_paths.append(path + next_path_candidates[j])
                best_sub_path = path + next_path_candidates[j+1]
            elif j == len(current_best_paths)-1:
                next_best_paths.append(
                    max(
                        best_sub_path,
                        path + next_path_candidates[j]
                    )
                )
                next_best_paths.append(path + next_path_candidates[j+1])
            else:
                next_best_paths.append(
                    max(
                        best_sub_path,
                        path + next_path_candidates[j]
                    )
                )
                best_sub_path = path + next_path_candidates[j+1]

        best_paths_dict[i+1] = next_best_paths

    return best_paths_dict, max(best_paths_dict[path_len-1])
```

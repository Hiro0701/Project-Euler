```python
import math
from tqdm import tqdm
```

```python
def find_divisor_num(n: int) -> int:
    divisor_set = set()
    sqrt_n = int(math.sqrt(n))

    for i in range(1, sqrt_n+1):
        if n % i == 0:
            divisor_set.add(i)
            divisor_set.add(n // i)

    return len(divisor_set)
```

```python
def find_triangle_num_by_divisors(n: int) -> int:
    current_num = 1
    
    with tqdm() as pbar:
        while True:
            if current_num % 2 == 1:
                a, b = current_num, (current_num + 1) // 2
            else:
                a, b = current_num // 2, current_num + 1

            divisor_num_a = find_divisor_num(a)
            divisor_num_b = find_divisor_num(b)

            if divisor_num_a * divisor_num_b > n:
                return a * b
            
            current_num += 1
            
            if current_num % 1000 == 0:
               pbar.update(1000)
               pbar.set_postfix(
                   a=a,
                   b=b,
                   a_b=a * b,
                   divisor_num=divisor_num_a * divisor_num_b,
               )
```

```python
triangle_num_5 = find_triangle_num_by_divisors(5)

print(triangle_num_5)
```

```python
triangle_num_500 = find_triangle_num_by_divisors(500)

print(triangle_num_500)
```

```python
# ~10,000 정도의 소수에 사용 가능
def find_nth_prime(n: int) -> int:
    assert n > 2
    
    idx = 2
    current_int = 3
    while True:
        past_int = 0
        for i in range(2, current_int+1):
            if (current_int+1) % i == 0:
                break
            past_int = i
        if past_int == (current_int):
            idx += 1
        current_int += 1
        if idx == n:
            return current_int
```

```python
print(find_nth_prime(10001))
```

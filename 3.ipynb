```python
def find_largest_prime_number(n):
    while True:
        for i in range(2, n):
            if n % i == 0:
                first_factor = i
                n = n // first_factor
                break
            if i == n-1:
                return n
```

```python
largest_prime_number = find_largest_prime_number(600851475143)

largest_prime_number
```
